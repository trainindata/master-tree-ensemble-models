In this notebook, we prepare the Bank Marketing dataset for the course.

The data comes from a direct marketing campaign of a Portuguese bank. The goal is to predict whether a client subscribes to a term deposit.

Here, we load the data from OpenML, encode the target as 0 and 1, give the variables descriptive names, and set the categorical variables to the category data type. Then, we save the data as a parquet file.

Dataset: [Bank Marketing](https://archive.ics.uci.edu/dataset/222/bank+marketing)

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml
```

```python
data = fetch_openml(
    name="bank-marketing",
    version=1,
    as_frame=True,
    parser="auto",
)

X = data.data.copy()
y = (data.target == "2").astype(int)
```

```python
X.columns = [
    "age",
    "job",
    "marital",
    "education",
    "default",
    "balance",
    "housing",
    "loan",
    "contact",
    "day",
    "month",
    "duration",
    "campaign",
    "pdays",
    "previous",
    "poutcome",
]
```

```python
X["pdays"] = X["pdays"].replace(-1, np.nan)
categorical_cols = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "poutcome",
]
X[categorical_cols] = X[categorical_cols].astype("category")
X.dtypes
```

```python
X.isnull().sum()
```

```python
df = pd.concat([X, y.rename("y")], axis=1)
df.to_parquet("bank_marketing.parquet")
df.to_csv("bank_marketing.csv", index=False)
```
